```python
img = cv2.imread("aaa.jpg")
plt.imshow(img)
```

```python
model = YOLO("./model.pt")
```

```python
results = model(img)[0]
```

```python
type(results)
```

```python
results.boxes
```

```python
for box in results.boxes.data.tolist():
    x1, y1, x2, y2, score, class_id = box
    x1, y1, x2, y2 = map(int, [x1, y1, x2, y2])

    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.putText(
        img,
        f"plate {score:.2f}",
        (x1, y1 - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        2,
    )
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.axis("off")
```

```python
plates = []

for box in results.boxes.data.tolist():
    x1, y1, x2, y2, score, class_id = box
    x1, y1, x2, y2 = map(int, [x1, y1, x2, y2])
    crop = img[y1:y2, x1:x2]
    plates.append(crop)

plt.imshow(cv2.cvtColor(plates[0], cv2.COLOR_BGR2RGB))
plt.axis("off")
```

```python
import torch
torch.__version__
```

```python
import easyocr

reader = easyocr.Reader(['en'], gpu=True)

for i, plate_crop in enumerate(plates):
       results = reader.readtext(plate_crop)

    print(f"--- Processing Plate {i+1} ---")

    for (bbox, text, prob) in results:
        if prob > 0.3:
            print(f"Detected Text: {text} (Confidence: {prob:.2f})")

            plt.figure(figsize=(2, 1))
            plt.imshow(cv2.cvtColor(plate_crop, cv2.COLOR_BGR2RGB))
            plt.title(f"Read: {text}")
            plt.axis("off")
            plt.show()
```
